```python
import pandas as pd
```

```python
PATH = '../data/raw/'

df = pd.read_csv(PATH + 'satellites.csv')
```

```python
df.head()
```

```python
print("Dataset shape:", df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("Duplicate NORAD IDs:", df["NORAD_CAT_ID"].duplicated().sum())

print("Unique satellite names:", df["OBJECT_NAME"].nunique())
```

```python
columns = ["MEAN_MOTION", "ECCENTRICITY", "INCLINATION", "RA_OF_ASC_NODE", "ARG_OF_PERICENTER","MEAN_ANOMALY", "BSTAR"]

display(df[columns].describe().T)
```

```python
df["EPOCH"] = pd.to_datetime(df["EPOCH"], utc=True, errors="coerce")

print("Invalid epochs:", df["EPOCH"].isna().sum())

print("Oldest epoch:", df["EPOCH"].min())
print("Newest epoch:", df["EPOCH"].max())
```

```python
reference_time = pd.Timestamp("2026-10-24 00:00:00", tz="UTC")

df["AGE_HOURS"] = (reference_time - df["EPOCH"]).dt.total_seconds() / 3600

print(df["AGE_HOURS"].describe())
```
