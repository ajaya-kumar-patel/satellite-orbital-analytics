```python
## Potential visiting vehicle
import json
from pathlib import Path

path = Path("../data/processed/iss/clean_candidate_satellites.json")

with path.open(encoding="utf-8") as f:
    records = json.load(f)["orbital_data"]

keywords = (
    "PROGRESS", "SOYUZ", "CREW DRAGON", "DRAGON",
    "CYGNUS", "HTV", "CST-", "STARLINER",
)

matches = [
    (r.get("NORAD_CAT_ID"), r.get("OBJECT_NAME"), r.get("EPOCH"))
    for r in records
    if any(
        keyword in str(r.get("OBJECT_NAME", "")).upper()
        for keyword in keywords
    )
]

print(f"Potential visiting-vehicle records remaining: {len(matches)}")
for norad_id, name, epoch in matches:
    print(f"{norad_id} | {name} | {epoch}")
```
